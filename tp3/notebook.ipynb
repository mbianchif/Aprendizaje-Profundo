## 1. Construya una red de Kohonen de 2 entradas que aprenda una distribución uniforme dentro del círculo unitario. Mostrar el mapa de preservación de topología. Probar con distribuciones uniformes dentro de otras figuras geométricas.

```python
from matplotlib.animation import FuncAnimation
from IPython.display import HTML
import matplotlib.pyplot as plt
import numpy as np


def animate_weights(X: np.ndarray, W_history: list[tuple[np.ndarray, ...]], interval: int = 100):
    fig, ax = plt.subplots(figsize=(6, 6))
    ax.scatter(X[:, 0], X[:, 1], color="C1", s=10)
    ax.set_aspect("equal")

    horiz_lines, = ax.plot([], [], color="C0", alpha=0.6)
    vert_lines, = ax.plot([], [], color="C0", alpha=0.6)

    def init():
        horiz_lines.set_data([], [])
        vert_lines.set_data([], [])
        return horiz_lines, vert_lines

    def update(frame):
        Wx, Wy = W_history[frame]
        R, C = Wx.shape
        nan_col = np.full((R, 1), np.nan)
        nan_row = np.full((C, 1), np.nan)

        Wx_h = np.hstack((Wx, nan_col)).ravel()
        Wy_h = np.hstack((Wy, nan_col)).ravel()
        horiz_lines.set_data(Wx_h, Wy_h)

        Wx_v = np.hstack((Wx.T, nan_row)).ravel()
        Wy_v = np.hstack((Wy.T, nan_row)).ravel()
        vert_lines.set_data(Wx_v, Wy_v)
        
        ax.set_title(f"Iteration {frame}")
        return horiz_lines, vert_lines

    ani = FuncAnimation(
        fig,
        update,
        frames=len(W_history),
        init_func=init,
        blit=True,
        interval=interval,
        repeat=False,
    )

    html = HTML(ani.to_jshtml())
    plt.close(fig)
    return html


SEED = 1234
```

```python
from src.distances import euclidean_distance
from src.utils import uniform_circle
from src.kohonen import Kohonen

X = uniform_circle(81, 1.0, seed=SEED)
m = Kohonen(2, (9, 9), euclidean_distance, seed=SEED)

weights = []
def tracker(m: Kohonen):
    ws = m.component_arrays()
    weights.append(ws)

m.train(X, eta0=2, max_iter=50, epoch_f=tracker)
```

```python
animate_weights(X, weights)
```

## 2. Resuelva (aproximadamente) el “Traveling salesman problem” para 200 ciudades con una red de Kohonen.

```python
from matplotlib.animation import FuncAnimation
from IPython.display import HTML
import matplotlib.pyplot as plt
import numpy as np


def animate_weights(X: np.ndarray, W_history: list[tuple[np.ndarray, ...]], interval: int = 100):
    fig, ax = plt.subplots(figsize=(6, 6))
    ax.scatter(X[:, 0], X[:, 1], color="C1", s=10)
    ax.set_aspect("equal")

    # DEFINE THE ARTISTS WITH THE CORRECT NAMES
    ring_lines, = ax.plot([], [], color="C0", alpha=0.6, marker='o', markersize=4)
    dummy_lines, = ax.plot([], [], color="none") # Keep for blitting safety

    def init():
        ring_lines.set_data([], [])
        dummy_lines.set_data([], [])
        return ring_lines, dummy_lines

    def update(frame):
        Wx, Wy = W_history[frame] 

        # --- 1. GUARANTEE 1D SHAPE (N,) ---
        # Using ravel() for maximum safety and ensuring a copy.
        Wx_flat = Wx.ravel().copy() 
        Wy_flat = Wy.ravel().copy()
        
        # --- 2. CLOSE THE LOOP (Size N+1) ---
        Wx_closed = np.append(Wx_flat, Wx_flat[0])
        Wy_closed = np.append(Wy_flat, Wy_flat[0])
        
        # --- 3. PLOT ---
        # FIX: Call set_data on the correctly named artist: ring_lines
        ring_lines.set_data(Wx_closed, Wy_closed) 
        
        # FIX: Update the dummy artist
        dummy_lines.set_data([], []) 

        ax.set_title(f"Iteration {frame}")
        
        # FIX: Return the correctly named artists
        return ring_lines, dummy_lines

    ani = FuncAnimation(
        fig,
        update,
        frames=len(W_history),
        init_func=init,
        blit=True,
        interval=interval,
        repeat=False,
    )

    html = HTML(ani.to_jshtml())
    plt.close(fig)
    return html


SEED = 1234
```

```python
from src.distances import chain_distance
from src.utils import uniform_square
from src.kohonen import Kohonen


n = 5
X = uniform_square(n, 1, seed=SEED)
m = Kohonen(2, (n,), chain_distance, seed=SEED)

i = 0
weights = []
def tracker(m: Kohonen):
    global i
    if i % 3 == 0:
        ws = m.component_arrays()
        weights.append(ws)
    i += 1

m.train(X, eta0=1, max_iter=50, epoch_f=tracker)
```

```python
animate_weights(X, weights)
```
