```python
from PIL import Image
from typing import Iterable
import os

def retrieve_images(path: str) -> Iterable[list[int]]:
    images = (
        Image.open(f"{path}/{file_path}").convert("1") for file_path in os.listdir(path)
    )

    for image in images:
        yield [1 if p else -1 for p in image.getdata()]
        image.close()
```

```python
from hopfield import Hopfield
from collections import defaultdict

images = retrieve_images("images")
sizes = defaultdict(list)

for image in images:
    sizes[len(image)].append(image)

for size, ps in sizes.items():
    m = Hopfield(size)
    for p in ps:
        m.train(p)

    for p in ps:
        print(m.learnt(p))

    
    
```
